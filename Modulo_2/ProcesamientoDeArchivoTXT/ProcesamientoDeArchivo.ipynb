```python
import pandas as pd
import numpy as np

# Leemos el archivo TXT indicando que:
# skiprows = Sirve para que se salte los primeros 18 renglones de texto informativo
# sep '\s' = Lo usamos para que los datos están separados por espacios en blanco
# names []= Le indicamos los nombres que usaremos para cada columna
# encoding = Utilizamos el formato 'latin-1' para que no fallen los acentos o símbolos
df = pd.read_csv('./Dataset/25001.txt', skiprows=18, sep='\s+', names=['Fecha', 'Precip', 'Evap', 'Tmax', 'Tmin'], encoding='latin-1')

# Mostramos un aviso de que todo salió bien y vemos los primeros 10 renglones
print("Archivo cargado con éxito")
df.head(10)
```

```python
# Quitamos la palabra "Nulo"
df = df.replace('Nulo', np.nan)

# Convertimos la columna de fecha a formato datetime
# coerce = Si encuentra basura o texto que no entiende, en lugar de marcar error, simplemente lo marca como NaT
df['Fecha'] = pd.to_datetime(df['Fecha'], errors='coerce')

# Se borran los elementos que no tengan una fecha valida o que sean NaT (Not at Time)
df = df.dropna(subset=['Fecha'])

# Convertimos las columnas de datos a números
cols_num = ['Precip', 'Evap', 'Tmax', 'Tmin']
df[cols_num] = df[cols_num].apply(pd.to_numeric, errors='coerce')

df.head(10) 
```
